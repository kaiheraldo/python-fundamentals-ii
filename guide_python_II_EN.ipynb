## 1 · Functions

A function is defined with `def`, receives **parameters** and gives back a result with `return`.

`print` only **shows** something on screen; `return` **hands back** the value so you can store it or use it.

````python
def rectangle_area(base, height):
    return base * height


result = rectangle_area(3, 4)  # the returned value is stored
print(result)
print(rectangle_area(10, 2) + 1)  # … or used directly
````

### Type hints and docstring
**Type hints** (`: float`, `-> float`) say which type of data each parameter expects and which type the function returns. They are optional and Python does not check them: they help whoever reads the code, and they let VS Code warn you.

The **docstring** is the text between triple quotes that explains the function. It is what `help()` shows.

````python
def rectangle_area(base: float, height: float) -> float:
    """Return the area of a rectangle."""
    return base * height


print(rectangle_area(3, 4))
help(rectangle_area)  # shows the first line and the docstring
````

### Default values and keyword arguments
If you do not pass an argument, its default value is used. You can also pass arguments by name, in any order.

````python
def power(base, exponent=2):
    return base ** exponent


print(power(5))                 # uses the default value: 5 ** 2
print(power(5, 3))              # by position: 5 ** 3
print(power(exponent=3, base=2))  # by name, in any order
````

### `*args`: any number of arguments
The function receives all the positional arguments as a **tuple**. It also works the other way round: if you already have a list, `*my_list` in the call spreads it into separate arguments.

````python
def add_all(*numbers):
    print("received:", numbers)
    return sum(numbers)


print(add_all(1, 2))
print(add_all(1, 2, 3, 4, 5))
print(add_all())

numbers = [10, 20, 30]
print(add_all(*numbers))  # the * spreads the list: same as add_all(10, 20, 30)
````

### `**kwargs`: any number of keyword arguments
The function receives all the keyword arguments as a **dictionary**.

With both together, `(*args, **kwargs)`, a function accepts any call: positional arguments go to `args` and keyword arguments go to `kwargs`. You will use this in decorators.

````python
def show_config(**options):
    for key, value in options.items():
        print(f"{key} = {value}")


show_config(model="gpt", temperature=0.7, language="en")


def log_call(*args, **kwargs):
    print("positional:", args, "| keyword:", kwargs)


log_call(1, 2, mode="fast")
````

### A function is just another value
It can be stored in a variable and passed to another function. `shout` without parentheses is the function; `shout("hello")` calls it. This idea is the basis of decorators.

````python
def shout(text):
    return text.upper() + "!"


def whisper(text):
    return text.lower() + "..."


def apply(function, text):
    return function(text)  # calls the function it was given


print(apply(shout, "Hello"))
print(apply(whisper, "Hello"))

my_function = shout  # no parentheses: we store the function, we do not call it
print(my_function("python"))
````

## 3 · Decorators

A **decorator** is a function that receives a function and returns another one that does the same **plus something extra**. First we need one earlier idea; then we build it in four steps.

### Before · A function can return another function
`make_greeter` creates another function inside, `inner`, and returns it **without parentheses**: it hands back the function, it does not run it. That is why the first two calls print nothing. It runs when you call it with `()`.

````python
def make_greeter(name):
    def inner():
        print(f"Hello, {name}")
    return inner  # no parentheses: returns the function, does not run it


greet_ana = make_greeter("Ana")    # prints nothing yet
greet_luis = make_greeter("Luis")

greet_ana()   # now it runs
greet_luis()
````

### Step 1 · A function that wraps another
It is the same as above, but now what `decorate` receives is a function. It creates `wrapper` (which calls the original) and returns it.

````python
def decorate(function):
    def wrapper():
        print("before")
        function()          # calls the original function
        print("after")
    return wrapper          # returns the new function


def say_hello():
    print("Hello")


decorated_hello = decorate(say_hello)  # we get the wrapped version
decorated_hello()
````

### Step 2 · The `@` is a shortcut
`@decorate` above the `def` is the same as `say_hello = decorate(say_hello)`.

````python
@decorate
def say_hello():
    print("Hello")


say_hello()
````

### Step 3 · Any arguments, and returning the result
The `wrapper` from step 1 accepts no arguments and returns nothing: with a function like `add(2, 3)` it would raise an error, and its result would be lost. To fix this, the wrapper receives `*args, **kwargs`, passes them to the original and **returns** its result.

````python
def decorate(function):
    def wrapper(*args, **kwargs):
        print("before")
        result = function(*args, **kwargs)  # passes on everything it receives
        print("after")
        return result                       # returns what the original returned
    return wrapper


@decorate
def add(a, b):
    return a + b


total = add(2, 3)
print("total:", total)
````

### Step 4 · `@wraps`
Without `@wraps`, the decorated function ends up being called `wrapper`. With it, the function keeps its name and its description (`__doc__`). Always include it.

````python
from functools import wraps


def decorate(function):
    @wraps(function)
    def wrapper(*args, **kwargs):
        print("before")
        result = function(*args, **kwargs)
        print("after")
        return result
    return wrapper


@decorate
def add(a, b):
    """Add two numbers."""
    return a + b


print(add.__name__)  # without @wraps it would say "wrapper"
print(add.__doc__)
````

### Exercise 5 · Stopwatch
Complete `timer` so that it measures how long the function takes. Inside `wrapper`:

1. `start = time.perf_counter()`
2. `result = function(*args, **kwargs)`
3. `seconds = time.perf_counter() - start`
4. Print `<name> took <seconds> s` using `function.__name__` and `f"{seconds:.2f}"`.
5. Return `result`.

**Expected output** (the time varies a little):
```
train took 0.30 s
model trained with 3 epochs
```

````python
import time
from functools import wraps


def timer(function):
    @wraps(function)
    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        result = function(*args, **kwargs)
        seconds = time.perf_counter() - start
        print(f"{function.__name__} took {seconds:.2f} s")
        return result
    return wrapper


@timer
def train(epochs):
    time.sleep(0.1 * epochs)  # simulates the work
    return f"model trained with {epochs} epochs"


print(train(3))
````

## 4 · Exceptions

When something fails, Python raises an **exception** and shows a **traceback**. Read it from the bottom up: the last line says what failed.

Run the cell: it fails on purpose.

````python
float("three")  # this fails on purpose
````

### `try` / `except`
The code that can fail goes in `try`; what you do if it fails goes in `except`. The program carries on. Always catch the **specific** error, never an empty `except:`.

````python
def convert(text):
    try:
        number = float(text)
    except ValueError:
        print(f"'{text}' is not a number, using 0")
        number = 0
    return number


print(convert("3.5"))
print(convert("three"))
print("the program carries on")
````

### Several `except`, `as e` and `finally`
You can write one `except` for each type of error. `as e` stores the error so you can show its message, and `finally` always runs, whether it failed or not.

````python
def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("cannot divide by zero")
        result = None
    except TypeError as e:
        print(f"wrong types: {e}")
        result = None
    finally:
        print("  (end of divide)")
    return result


print(divide(10, 2))
print(divide(10, 0))
print(divide(10, "two"))
````

### `raise`: raising an error yourself
If a value is not valid, do not carry on: raise an exception with `raise`.

````python
def percentage(value):
    if not 0 <= value <= 100:
        raise ValueError(f"the percentage must be between 0 and 100, not {value}")
    return f"{value} %"


print(percentage(42))
try:
    print(percentage(150))
except ValueError as e:
    print("Error:", e)
````

### Your own exceptions
An exception of your own is a class that inherits from `Exception`; it needs nothing else inside. This way you can catch only your errors and not Python's.

````python
class InsufficientBalanceError(Exception):
    """There is not enough balance for the operation."""


def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientBalanceError(f"you want {amount} and there is only {balance}")
    return balance - amount


try:
    print(withdraw(100, 30))
    print(withdraw(100, 300))
except InsufficientBalanceError as e:
    print("Operation cancelled:", e)
````

### Exercise 7 · Your own exception

### Step 1 · The module
The cell creates `playlist.py`. The songs are in a dictionary: title → seconds. Complete:

- `add_song(title, seconds)`: stores the song in `songs`.
- `find(title)`: if it is not there, it raises `SongNotFoundError` with the message `"<title>" is not in the playlist`; if it is there, it returns its seconds.
- `total_duration(*titles)`: returns the sum of the seconds of all the titles (use `find`).

````python
%%writefile playlist.py
"""A very simple playlist."""


class SongNotFoundError(Exception):
    """The requested song is not in the playlist."""


songs = {
    "Bohemian Rhapsody": 354,
    "Despacito": 229,
}


def add_song(title, seconds):
    songs[title] = seconds


def find(title):
    if title not in songs:
        raise SongNotFoundError(f'"{title}" is not in the playlist')
    return songs[title]


def total_duration(*titles):
    return sum(find(title) for title in titles)
````

### Step 2 · The decorator
Complete `announce`: it prints `♪ Now playing…`, calls the function and returns its result. `play` is already written.

**Expected output:**
```
♪ Now playing…
Bohemian Rhapsody (5:54)
```

````python
from functools import wraps
import playlist

def announce(function):
    @wraps(function)
    def wrapper(*args, **kwargs):
        print("♪ Now playing…")
        return function(*args, **kwargs)
    return wrapper


@announce
def play(title):
    seconds = playlist.find(title)
    return f"{title} ({seconds // 60}:{seconds % 60:02d})"


print(play("Bohemian Rhapsody"))
````
